QSorter

```python
def qsorter(arr):
    if len(arr) <= 1:
        return arr
    
    pivot = arr[len(arr) //2]
    left = [x for x in arr if x < pivot]
    mid = [x for x in arr if x == pivot] 
    right = [x for x in arr if x > pivot]
    
    return qsorter(left) + mid + qsorter(right)

print( qsorter( [2,1,1,0] ) )
```

Binary Search

```python
lst = (1,2,3,4,5,6,7)
k = 2

for x in range(len(lst)):
    if k == lst[x]:
        is_found = True
        print(f"{k} has an index of {x}")
        break
    
if is_found == False:
    print(f"{k} was not found in the list")
```
